```python
%%capture
import os
import pandas as pd
import numpy as np
from dj_notebook import activate
from pathlib import Path

env_file = os.environ["INTECOMM_ENV"]
analysis_folder = Path(os.environ["INTECOMM_ANALYSIS_FOLDER"])
reports_folder = Path(os.environ["INTECOMM_ANALYSIS_FOLDER"])
plus = activate(dotenv_file=env_file)
```

```python
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
from intecomm_analytics.dataframes import get_df_main_1858
```

```python
df_main = get_df_main_1858(None)
```

```python
df_main["gender"].value_counts()
```

```python
df_main["htn"] = df_main["htn"].astype("Int64")
df_main["bp_controlled_baseline"] = df_main["bp_controlled_baseline"].fillna(0)
df_main["bp_controlled_baseline"] = df_main["bp_controlled_baseline"].astype("Int64")
logit_model = sm.Logit.from_formula("htn ~ bp_controlled_baseline", data=df_main)
result = logit_model.fit()
print(result.summary())
```

```python
fitted_probabilities = result.predict(df_main)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x=df_main['bp_controlled_baseline'], y=fitted_probabilities)
plt.xlabel('bp_controlled_baseline')
plt.ylabel('Predicted Probability')
plt.title('Predicted Probabilities vs bp_controlled_baseline')
plt.show()
```

```python
df_htn = df_main[df_main.htn_only==1].copy()
df_htn["primary_controlled"] = df_htn["primary_controlled"].fillna(0)
df_htn["gender"] = df_htn["gender"].map({"Male": 0, "Female": 1})
df_htn["gender"] = df_htn["gender"].astype("Int64")
df_htn["smoking_status"] = df_htn["smoking_status"].map({"nonsmoker": 0, "former_smoker": 1, "smoker": 1})
df_htn["smoking_status"] = df_htn["smoking_status"].astype("Int64")
df_htn["smoking_status"] = df_htn["smoking_status"].fillna(1)
```

```python
logit_model = sm.Logit.from_formula("htn ~ bp_controlled_baseline", data=df_htn)
result = logit_model.fit()
print(result.summary())
```

```python
logit_model = sm.Logit.from_formula("primary_controlled ~ age_in_years", data=df_htn)
result = logit_model.fit()
print(result.summary())
```

```python
fitted_probabilities = result.predict(df_htn)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x=df_htn['age_in_years'], y=fitted_probabilities)
plt.xlabel('age_in_years')
plt.ylabel('Predicted Probability')
plt.title('Predicted Probabilities vs age_in_years')
plt.show()
```

```python
logit_model = sm.Logit.from_formula("primary_controlled ~ smoking_status", data=df_htn)
result = logit_model.fit()
print(result.summary())
```
