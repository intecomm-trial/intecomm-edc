```python
%%capture
import os
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
# import seaborn as sns
import scipy.stats as stats
from datetime import datetime
from django_pandas.io import read_frame
from pathlib import Path


from dj_notebook import activate

# pd.options.mode.copy_on_write = True
# pd.options.mode.chained_assignment = "raise"
env_file = os.environ["INTECOMM_ENV"]
documents_folder = os.environ["INTECOMM_DOCUMENTS_FOLDER"]
plus = activate(dotenv_file=env_file)

report_folder = Path(documents_folder)
```

```python
from intecomm_analytics.dataframes import get_patientlog_df

age_bins1 = {"<18": (0,17),"18-29": (18, 29),"30-39": (30,39), "40-49":(40,49),"50-59": (50,59),"60-69":(60,69),"70-79":(70,79),"80-89": (80, 89),"90-99": (90, 99),"100+": (100, 109)}
interval_ix1 = pd.IntervalIndex.from_tuples(age_bins1.values())

df_pat = get_patientlog_df()

df_cut1 = pd.cut(df_pat["age_in_years"], interval_ix1).value_counts(sort=False, normalize=True).to_frame().reset_index()

df_cut1.columns = ["interval", "proportion"]

age_bins_reversed1 = {pd.Interval(v[0], v[1]): k for k, v in age_bins1.items()}

df_cut1["age_in_years"] =  df_cut1["interval"].apply(lambda x: age_bins_reversed1.get(x))
```

```python
age_bins2 = {"<40": (0,39),"40-49":(40,49),"50-59": (50,59),"60-69": (60,69),"70+":(70,100)}
interval_ix2 = pd.IntervalIndex.from_tuples(age_bins2.values())

df_cut2 = pd.cut(df_pat["age_in_years"], interval_ix2).value_counts(sort=False, normalize=True).to_frame().reset_index()

df_cut2.columns = ["interval", "proportion"]

age_bins_reversed2 = {pd.Interval(v[0], v[1]): k for k, v in age_bins2.items()}

df_cut2["age_in_years"] =  df_cut2["interval"].apply(lambda x: age_bins_reversed2.get(x))
```
